```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

url="https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df=pd.read_csv(url)

print(df.head())
```

```python
print("Rows and Columns:",df.shape)
print("\nColumn Names:")
print(df.columns)
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
print(df.describe())
```

```python
print(df["Survived"].value_counts())

sns.countplot(x="Survived",data=df)
plt.title("Survival Count")
plt.show()
```

```python
sns.countplot(x="Sex",hue="Survived",data=df)
plt.title("Survival by Gender")
plt.show()
```

```python
sns.histplot(data=df,x="Age",bins=20)
plt.title("Age Distribution")
plt.show()
```

```python
sns.countplot(x="Pclass",hue="Survived",data=df)
plt.title("Survival by Passenger Class")
plt.show()
```



Key Insights

1. The dataset contains information about Titanic passengers.
2. Survival rates were different between male and female passengers.
3. Passenger class had an effect on survival.
4. The dataset contains passengers from different age groups.
5. Data visualization helps identify patterns in passenger survival.
