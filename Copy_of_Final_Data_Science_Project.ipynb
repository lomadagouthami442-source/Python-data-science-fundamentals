```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data={
"Student":["A","B","C","D","E","F","G","H","I","J"],
"Hours":[2,3,4,5,6,7,8,9,5,7],
"Attendance":[70,75,80,85,90,92,95,98,82,94],
"Marks":[45,50,58,65,72,78,85,92,68,80]
}
df=pd.DataFrame(data)
print(df)
```

```python
print("Shape:",df.shape)
print("\nColumns:")
print(df.columns)
print("\nFirst 5 rows:")
print(df.head())
```

```python
print(df.isnull().sum())
```

```python
print(df.describe())
```

```python
print("Highest Marks:",df["Marks"].max())
print("Lowest Marks:",df["Marks"].min())
print("Average Marks:",round(df["Marks"].mean(),2))
```

```python
plt.bar(df["Student"],df["Marks"])
plt.title("Student Marks")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.show()
```

```python
plt.plot(df["Student"],df["Marks"],marker="o")
plt.title("Student Marks Trend")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.show()
```

```python
sns.scatterplot(x=df["Hours"],y=df["Marks"])
plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.show()
```
