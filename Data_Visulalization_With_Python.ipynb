```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data={
"Student":["A","B","C","D","E","F","G","H"],
"Marks":[85,72,90,65,78,88,92,75],
"Attendance":[90,82,95,70,85,91,96,80]
}
df=pd.DataFrame(data)
df
```

```python
plt.figure(figsize=(8,5))
plt.bar(df["Student"],df["Marks"])
plt.title("Student Marks")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.plot(df["Student"],df["Marks"],marker="o")
plt.title("Student Marks Trend")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.show()
```

```python
plt.figure(figsize=(7,7))
plt.pie(df["Marks"],labels=df["Student"],autopct="%1.1f%%")
plt.title("Marks Distribution")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["Marks"],bins=5)
plt.title("Distribution of Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df["Attendance"],df["Marks"])
plt.title("Attendance vs Marks")
plt.xlabel("Attendance")
plt.ylabel("Marks")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x=df["Marks"])
plt.title("Marks Distribution using Seaborn")
plt.show()
```

The visualizations show that students with higher attendance generally have higher marks. The bar and line charts compare student performance, while the histogram and box plot show the distribution of marks. The scatter plot helps identify the relationship between attendance and marks.
