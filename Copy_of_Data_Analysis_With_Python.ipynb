```python
import pandas as pd
import numpy as np
```

```python
data={
    "Name":["Asha","Ravi","Priya","Kiran","Rahul"],
    "Age":[20,21,20,22,21],
    "Marks":[85,78,92,65,88],
    "Department":["CSE","ECE","CSE","EEE","CSE"]
}
df=pd.DataFrame(data)
print(df)
```

```python
print(df.head())
```

```python
print(df.info())
```

```python
print(df.describe())
```

```python
print(df.isnull().sum())
```

```python
df=df.drop_duplicates()
print(df)
```

```python
high_scores=df[df["Marks"]>80]
print(high_scores)
```

```python
sorted_data=df.sort_values(by="Marks",ascending=False)
print(sorted_data)
```

```python
print("Average Marks:",df["Marks"].mean())
print("Highest Marks:",df["Marks"].max())
print("Lowest Marks:",df["Marks"].min())
```

```python
print(df.groupby("Department")["Marks"].mean())
```

Conclusion

The student dataset was successfully explored and analyzed using Python, Pandas, and NumPy.

The analysis included data inspection, missing-value checking, duplicate removal, filtering, sorting, grouping, and statistical calculations.
